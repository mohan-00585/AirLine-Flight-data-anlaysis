CHARTS

```python
#Bar plot
plt.figure(figsize=(10, 5))

sns.barplot(data=df_group2,x="duration",y="airline",hue="airline",palette="Set2",legend=False)
plt.title("Average Travel Duration by Airline",fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Average Duration (Hours)", fontsize=12)
plt.ylabel("Airline", fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.tight_layout()
plt.show()
```

```python
#scatter 
plt.figure(figsize=(8, 5))

data_groupby=df.groupby('airline')[['duration','price']].value_counts().nlargest(50).reset_index()
sns.scatterplot(data=data_groupby, x="duration", y="price",hue="airline", palette="tab10", alpha=0.4, s=100)
plt.title("Travel Duration vs Ticket Price",fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Travel Duration (Hours)", fontsize=12)
plt.ylabel("Ticket Price (INR)", fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.legend(title="Airline", fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#pie
df_data = df["destination_city"].value_counts()

plt.figure(figsize=(8, 6))
plt.pie(df_data,labels=df_data.index,autopct="%1.1f%%",colors=sns.color_palette("Pastel1"),startangle=90,
        textprops={"fontsize": 11},
        wedgeprops={"edgecolor": "white"})
plt.title("Flight Options by Destination City",
          fontsize=16, fontweight="bold", color="navy", pad=20)
plt.tight_layout()
plt.show()
```

```python
#histogram 
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="distance_km", bins=15,color="coral", edgecolor="white")
plt.title("Distance Distribution of Flight Options",fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Distance (km)", fontsize=12)
plt.ylabel("Number of Flight Options", fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
#volinpot 
plt.figure(figsize=(9, 5))

sns.violinplot(data=df, x="stops", y="duration", hue="stops",order=["zero", "one", "two_or_more"],
               palette="Set3", legend=False, cut=0)
plt.title("Travel Duration by Number of Stops",fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Number of Stops", fontsize=12)
plt.ylabel("Travel Duration (Hours)", fontsize=12)
plt.xticks([0, 1, 2], ["Non-stop", "1 Stop", "2+ Stops"], fontsize=11)
plt.yticks(fontsize=11)
plt.ylim(bottom=0)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#stem
df_stem = df.groupby("arrival_time")["price"].mean()

df_stem = df_stem.reindex(
    ["Early_Morning", "Morning", "Afternoon",
     "Evening", "Night", "Late_Night"]
).reset_index()
plt.figure(figsize=(10, 5))
plt.stem(df_stem["arrival_time"].str.replace("_", " "),df_stem["price"],linefmt="m-", markerfmt="ro")
plt.title("Ticket Price by Arrival Time",fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Arrival Time", fontsize=12)
plt.ylabel("Average Ticket Price (INR)", fontsize=12)
plt.xticks(rotation=20, fontsize=11)
plt.yticks(fontsize=11)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#Boxplot
plt.figure(figsize=(9, 5))

sns.boxplot(data=df, x="class", y="price", hue="class",order=["Economy", "Business"],
            palette="coolwarm",legend=False)
plt.title("Ticket Price Distribution by Travel Class",
          fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Travel Class", fontsize=12)
plt.ylabel("Ticket Price (INR)", fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#stackplot
df_stack = df.groupby(["departure_time", "source_city"]).size().unstack(fill_value=0)

df_stack = df_stack.reindex(["Early_Morning", "Morning", "Afternoon",
                             "Evening", "Night", "Late_Night"])
plt.figure(figsize=(10, 5))
plt.stackplot(range(len(df_stack)),df_stack["Delhi"],df_stack["Mumbai"],labels=["Delhi", "Mumbai"],
    colors=["gold", "mediumseagreen"],step="mid",alpha=0.8)
plt.title("Delhi and Mumbai Flight Options by Departure Time",
          fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Departure Time", fontsize=12)
plt.ylabel("Number of Flight Options", fontsize=12)
plt.xticks(range(len(df_stack)),df_stack.index.str.replace("_", " "),
    rotation=20,
    fontsize=11
)
plt.yticks(fontsize=11)
plt.legend(title="Source City", fontsize=11)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#step plot
df_step = df.groupby("distance_km")["duration"].mean().reset_index()

df_step = df_step.sort_values("distance_km")
plt.figure(figsize=(10, 5))
plt.step(df_step["distance_km"], df_step["duration"],
         where="mid", color="royalblue", linewidth=2, marker="o")
plt.title("Average Travel Duration by Distance",
          fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Distance (km)", fontsize=12)
plt.ylabel("Average Duration (Hours)", fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#KDE plot
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df, x="price",fill=True, cut=0, color="darkviolet")
plt.title("Ticket Price Density",fontsize=16, fontweight="bold", color="navy")
plt.xlabel("Ticket Price (INR)", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
plt.grid(alpha=0.8)
plt.tight_layout()
plt.show()
```

```python
sns.set_style("whitegrid")

time_order = ["Early_Morning", "Morning", "Afternoon",
              "Evening", "Night", "Late_Night"]
fig1, axes = plt.subplots(2, 2, figsize=(16, 12), facecolor="aliceblue")
fig1.suptitle("Airline Network and Travel Choices",
              fontsize=22, fontweight="bold", color="indigo")

data1 = df.groupby("airline")["distance_km"].mean().sort_values()

axes[0,0].bar(data1.index.str.replace("_", " "),data1.values,color="cornflowerblue",width=0.6)
axes[0,0].bar_label(axes[0,0].containers[0],fmt="%.1f",padding=4,fontsize=10)
axes[0,0].set_title("Average Distance by Airline",fontsize=14, fontweight="bold", color="indigo")
axes[0,0].set_xlabel("Airline", fontsize=11)
axes[0,0].set_ylabel("Average Distance (km)", fontsize=11)
axes[0,0].tick_params(axis="x", rotation=20)


data2 = df["stops"].value_counts().reindex(["zero", "one", "two_or_more"])

axes[0,1].pie(data2,labels=["Non-stop", "1 Stop", "2+ Stops"],autopct="%1.1f%%",startangle=90,
    pctdistance=0.8,colors=["mediumturquoise", "lightsalmon", "plum"],wedgeprops={"width": 0.4, "edgecolor": "white"},
    textprops={"fontsize": 11})
axes[0,1].text(0, 0,f"{len(df):,}\nFlight options",ha="center",va="center",fontsize=14,color="indigo")
axes[0,1].set_title("Flight Options by Number of Stops",fontsize=14, fontweight="bold", color="indigo")


data3 = df.groupby("departure_time")["price"].mean().reindex(time_order)

axes[1,0].plot(data3.index.str.replace("_", " "),data3.values,color="darkorange",marker="o",linewidth=2)
axes[1,0].set_title("Average Ticket Price by Departure Time",fontsize=14, fontweight="bold", color="indigo")
axes[1,0].set_xlabel("Departure Time", fontsize=11)
axes[1,0].set_ylabel("Average Ticket Price (INR)", fontsize=11)
axes[1,0].set_ylim(bottom=0)
axes[1,0].tick_params(axis="x", rotation=20)


data4 = df.groupby(["source_city", "destination_city"]).size().unstack(fill_value=0)

sns.heatmap(data4,annot=True,fmt="d",cmap="YlGnBu",linewidths=0.5,cbar_kws={"label": "Flight Options"},
    ax=axes[1,1]
)

axes[1,1].set_title("Flight Options by Source and Destination",fontsize=14, fontweight="bold", color="indigo")
axes[1,1].set_xlabel("Destination City", fontsize=11)
axes[1,1].set_ylabel("Source City", fontsize=11)
axes[1,1].tick_params(axis="x", rotation=20)
axes[1,1].tick_params(axis="y", rotation=0)

plt.tight_layout()
plt.show()
```

```python
sns.set_style("whitegrid")

time_order = ["Early_Morning", "Morning", "Afternoon",
              "Evening", "Night", "Late_Night"]
df_group4 = df.groupby(["airline", "class"])["price"].mean().reset_index()
df_group6 = df.groupby(["arrival_time", "stops"]).size().unstack(fill_value=0)
df_group6 = df_group6.reindex(time_order)


fig2, axes = plt.subplots(2, 2, figsize=(16, 12), facecolor="floralwhite")
fig2.suptitle("Ticket Prices and Journey Patterns",fontsize=22, fontweight="bold", color="darkred")


sns.barplot(data=df_group4,x="airline",y="price",hue="class",hue_order=["Economy", "Business"],
    palette=["royalblue", "coral"],ax=axes[0,0])
for bars in axes[0,0].containers:
    axes[0,0].bar_label(bars, fmt="%.0f", padding=4, fontsize=10)
axes[0,0].set_title("Average Ticket Price by Airline and Class",fontsize=14, fontweight="bold", color="darkred")
axes[0,0].set_xlabel("Airline", fontsize=11)
axes[0,0].set_ylabel("Average Ticket Price (INR)", fontsize=11)
axes[0,0].set_ylim(0, df_group4["price"].max() * 1.2)
axes[0,0].tick_params(axis="x", rotation=20)
axes[0,0].legend(title="Travel Class", loc="upper left", fontsize=10)


sns.scatterplot(x=df["duration"],y= df["price"]/df["distance_km"],palette="darkorange",alpha=0.25,
    s=14,ax=axes[0,1])
axes[0,1].set_title("Journey Duration vs Ticket Price per km",fontsize=14, fontweight="bold", color="darkred")
axes[0,1].set_xlabel("Journey Duration (Hours)", fontsize=11)
axes[0,1].set_ylabel("Ticket Price per km (INR/km)", fontsize=11)


sns.boxplot(data=df,x="destination_city",y="distance_km",color="lightseagreen",ax=axes[1,0])
axes[1,0].set_title("Distance Distribution by Destination City",fontsize=14, fontweight="bold", color="darkred")
axes[1,0].set_xlabel("Destination City", fontsize=11)
axes[1,0].set_ylabel("Distance (km)", fontsize=11)
axes[1,0].tick_params(axis="x", rotation=20)


axes[1,1].stackplot(range(len(df_group6)),df_group6["zero"],df_group6["one"] + df_group6["two_or_more"],
    labels=["Non-stop", "Connecting"],colors=["mediumpurple", "gold"],step="mid",alpha=0.8)
axes[1,1].set_title("Non-stop and Connecting Options by Arrival Time",fontsize=14, fontweight="bold", color="darkred")
axes[1,1].set_xlabel("Arrival Time", fontsize=11)
axes[1,1].set_ylabel("Number of Flight Options", fontsize=11)
axes[1,1].set_xticks(range(len(df_group6)))
axes[1,1].set_xticklabels(df_group6.index.str.replace("_", " "),rotation=20)
axes[1,1].legend(title="Journey Type", fontsize=10)

plt.tight_layout(rect=[0, 0, 1, 0.95], h_pad=3, w_pad=3)
plt.show()
```

```python
sns.set_style("whitegrid")

time_order = ["Early_Morning", "Morning", "Afternoon",
              "Evening", "Night", "Late_Night"]

data5 = df.groupby(["airline", "flight"]).size().reset_index(name="flight_options")
data7 = df.groupby(["source_city", "destination_city"]).size().reset_index(name="flight_options")

fig3, axes = plt.subplots(2, 3, figsize=(22, 12), facecolor="white",)
fig3.suptitle("Journey Timing and Flight Option Coverage",fontsize=22, fontweight="bold", color="teal")

sns.violinplot(data=df,x="departure_time",y="duration",order=time_order,color="lightskyblue",cut=0,ax=axes[0,0])
axes[0,0].set_title("Journey Duration by Departure Time",fontsize=14, fontweight="bold", color="teal")
axes[0,0].set_xlabel("Departure Time", fontsize=11)
axes[0,0].set_ylabel("Journey Duration (Hours)", fontsize=11)
axes[0,0].set_xticklabels(time_order,rotation=20)


sns.stripplot(data=data5,x="airline",y="flight_options",color="mediumorchid",alpha=0.5,size=3,jitter=0.2,
    ax=axes[0,1])
axes[0,1].set_title("Flight Options per Flight Code by Airline",fontsize=14, fontweight="bold", color="teal")
axes[0,1].set_xlabel("Airline", fontsize=11)
axes[0,1].set_ylabel("Flight Options per Flight Code", fontsize=11)
axes[0,1].tick_params(axis="x", rotation=20)


sns.kdeplot(x=df["price"] / df["distance_km"],fill=True,cut=0,color="darkviolet",ax=axes[0,2])
axes[0,2].set_title("Ticket Price per Kilometre Distribution",fontsize=14, fontweight="bold", color="teal")
axes[0,2].set_xlabel("Ticket Price per km (INR/km)", fontsize=11)
axes[0,2].set_ylabel("Density", fontsize=11)


axes[1,0].hist(df["duration"],bins=12,color="coral",edgecolor="white")
axes[1,0].bar_label(axes[1,0].containers[0],fmt="%.0f",padding=3,fontsize=9)
axes[1,0].set_title("Journey Duration Distribution",fontsize=14, fontweight="bold", color="teal")
axes[1,0].set_xlabel("Journey Duration (Hours)", fontsize=11)
axes[1,0].set_ylabel("Number of Flight Options", fontsize=11)
axes[1,0].margins(y=0.18)


sns.countplot(data=df,x="source_city",order=df["source_city"].value_counts().index,color="seagreen",ax=axes[1,1])
axes[1,1].bar_label(axes[1,1].containers[0],fmt="%.0f",padding=4,fontsize=10)
axes[1,1].set_title("Flight Options by Source City",fontsize=14, fontweight="bold", color="teal")
axes[1,1].set_xlabel("Source City", fontsize=11)
axes[1,1].set_ylabel("Number of Flight Options", fontsize=11)
axes[1,1].tick_params(axis="x", rotation=20)
axes[1,1].margins(y=0.18)


sns.ecdfplot(data=data7,x="flight_options",stat="percent",color="darkgreen",linewidth=2,ax=axes[1,2])
axes[1,2].set_title("Flight Option Counts Across Routes",fontsize=14, fontweight="bold", color="teal")
axes[1,2].set_xlabel("Flight Options per Route", fontsize=11)
axes[1,2].set_ylabel("Routes at or Below Count (%)", fontsize=11)
axes[1,2].set_ylim(0, 100)
axes[1,2].set_yticks([0, 25, 50, 75, 100])

plt.tight_layout(rect=[0, 0, 1, 0.95], h_pad=3, w_pad=3)
plt.show()
```
